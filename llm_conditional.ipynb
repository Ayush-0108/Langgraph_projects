```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from typing import TypedDict, Annotated,Literal
from pydantic import Field,BaseModel
from dotenv import load_dotenv
import operator
```

```python
load_dotenv()
```

```python
model = ChatGroq(model="openai/gpt-oss-20b")
```

```python
class SentimentSchema(BaseModel):
    sentiment: Literal['Positive','Negative'] = Field(description="Sentiment of the review")
```

```python
class DiagnosisSchema(BaseModel):
    issue_type: str=Field(description="Category of the issue type of the review")
    tone: Literal['angry','frustrated','disappointed','calm'] = Field(description='emotional tone expressed by the reviewer')
    urgency: Literal['low',"medium",'high']=Field(description="How urgent the issue seems to be")
```

```python
structured_model1 = model.with_structured_output(SentimentSchema)
structured_model2 = model.with_structured_output(DiagnosisSchema)
```

```python
class ReviewState(TypedDict):
    review:str
    Sentiment: Literal['Positive','Negative']
    diagnosis: dict
    response: str
```
